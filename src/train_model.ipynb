```python
import pandas as pd
from sklearn.cluster import KMeans
import joblib
```

```python
# Load processed data
df = pd.read_csv("../data/processed/processed_crime_data.csv")
```

```python
# Features only
X = df.drop(columns=[df.columns[-1]])
```

```python
# Train KMeans model
kmeans = KMeans(n_clusters=3, random_state=42)
df["hotspot_level"] = kmeans.fit_predict(X)
```

```python
# Save model
joblib.dump(kmeans, "../models/hotspot_model.pkl")
```

```python
# Save predictions
df.to_csv("../data/processed/hotspot_predictions.csv", index=False)
```

```python
print("✅ Hotspot model trained and saved successfully")
```
