```python
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("../data/raw/crime_dataset_india.csv")
```

```python
# Remove missing values
df = df.dropna()
```

```python
# Normalize column names
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
# Encode categorical columns
categorical_columns = ["state", "city", "crime_type"]

for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].astype("category").cat.codes
```

```python
# Select features
feature_columns = []
for col in ["state", "city", "crime_type", "latitude", "longitude"]:
    if col in df.columns:
        feature_columns.append(col)
```

```python
# Select target
target_column = "crime_count" if "crime_count" in df.columns else df.columns[-1]

processed_df = df[feature_columns + [target_column]]
```

```python
# Save processed dataset
processed_df.to_csv(
    "../data/processed/processed_crime_data.csv",
    index=False
)
```
